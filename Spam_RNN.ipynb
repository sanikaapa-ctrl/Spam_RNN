```python
# 1. Reload clean dataset
df = pd.read_csv("spam.csv", encoding='latin-1')
df = df.dropna(subset=['v1', 'v2'])


df['v1'] = df['v1'].astype(str).str.strip().map({'ham': 0, 'spam': 1})
df = df.dropna(subset=['v1'])


tokenizer = Tokenizer()
tokenizer.fit_on_texts(df['v2'])
sequences = tokenizer.texts_to_sequences(df['v2'])
max_len = 50
X_cleaned = pad_sequences(sequences, maxlen=max_len)
y_cleaned = df['v1'].values.astype('float32')


X_train, X_test, y_train, y_test = train_test_split(
    X_cleaned, y_cleaned,
    test_size=0.2,
    random_state=32,
    stratify=y_cleaned
)


model.fit(
    X_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.01
)


loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
test = ["today we have a offer"]


test_seq = tokenizer.texts_to_sequences(test)
test_pad = pad_sequences(test_seq, maxlen=max_len)

prediction = model.predict(test_pad)
print (prediction[0][0])
if prediction[0][0] <= 0.5:
    print("Spam")
else:
    print("Ham (Normal Message)")
```
